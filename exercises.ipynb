```python
#to check a palindrome
text = input("Enter a word: ")
clean_text = text[::-1]

if clean_text == text:
    print(text, 'is a palindrome.')
else:
    print(text, 'is not a palindrome.')
```

```python
#check a palindrome that ignores spaces and upper case
text = input("Enter a word or sentence: ")
clean_text = text[::-1].lower().replace(" ", "")

if clean_text == text:
    print(text, 'is a palindrome.')
else:
    print(text, 'is not a palindrome.')
```

```python
#check factorial
n = int(input("Enter a number: "))
factorial = 1

for i in range(1, n + 1):
    factorial *= i

print("Factorial from 1 to", n, "is:", factorial)
```

```python
#check prime number
n = int(input("Enter N: "))

for num in range(2, n + 1):
    is_prime = True

    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            is_prime = False
            break

    if is_prime:
        print(num)

```

```python
#generate a fibonacci series
n = int(input("Enter N: "))
a, b = 0, 1
count = 0

while count < n:
    print(a, end=" ")
    a, b = b, a + b
    count += 1
```

```python
def find_duplicates(text):
    counts = {}
    for char in text:
        counts[char] = counts.get(char, 0) + 1
    return {char: n for char, n in counts.items() if n > 1}
print(find_duplicates("consonants"))
```

```python
def sum_even(num):
    even_sum = 0
    odd_sum = 0
    for i in range(1, num + 1):
        if i % 2 == 0:
            even_sum += i
        else:
            odd_sum += i
    return even_sum
sum_even(10)
```

```python
#function which returns the slope of a linear equation
def calculate_slope(x1, y1, x2, y2):
    if x2 == x1:
        raise ValueError("Slope is undefined for vertical line.")
    else:
        return (y2 - y1) / (x2 - x1)
calculate_slope(1, 2, 5, 18)
```
